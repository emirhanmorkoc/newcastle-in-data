```python
import pandas as pd
import sqlite3

df = pd.read_csv("../data/clean/newcastle_clean.csv", parse_dates=["date"])
df["date"] = df["date"].dt.strftime("%Y-%m-%d")  # store dates as text for SQLite

con = sqlite3.connect("../data/clean/newcastle.db")
df.to_sql("sales", con, if_exists="replace", index=False)

print(pd.read_sql("SELECT COUNT(*) AS n_rows FROM sales", con))
print(pd.read_sql("SELECT * FROM sales LIMIT 3", con))
con.close()
```
