```python
import pandas as pd, glob

cols = ["transaction_id","price","date","postcode","property_type",
        "new_build","tenure","paon","saon","street","locality",
        "town","district","county","ppd_category","record_status"]

files = sorted(glob.glob("../data/raw/*.csv"))
print(len(files), "dosya bulundu:")
for f in files: print(" ", f)

parts = []
for f in files:
    for chunk in pd.read_csv(f, names=cols, header=None, chunksize=500_000):
        parts.append(chunk[chunk["district"] == "NEWCASTLE UPON TYNE"])

nc = pd.concat(parts, ignore_index=True)
nc["date"] = pd.to_datetime(nc["date"])

print(nc.shape)
print(nc["date"].min(), nc["date"].max())
print(nc.groupby(nc["date"].dt.year).size())

nc.to_csv("../data/clean/newcastle_raw_filtered.csv", index=False)
print("Kaydedildi")
```
